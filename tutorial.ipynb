```python
import numpy as np
from astropy.io import fits
import os 
import matplotlib.pyplot as plt
import glob
```

```python
## read in data 
path = "/Users/U8019618/Documents/teaching/2026/PHY5001/Assessment4_files/"

## brightest stars 
brightest_stars = np.genfromtxt(path+'Brightest_stars_GAIA_Report.csv', delimiter=',', names=True, dtype=None, encoding='utf-8')

## nearest stars
nearest_stars = np.genfromtxt(path+'Nearest_stars_GAIA_Report.csv', delimiter=',', names=True, dtype=None, encoding='utf-8')

## ngc2477 
ngc2477 = np.genfromtxt(path+'NGC_2477.csv', delimiter=',', names=True, dtype=None, encoding='utf-8')

## ngc2682
ngc2682 = np.genfromtxt(path+'NGC_2682.csv', delimiter=',', names=True, dtype=None, encoding='utf-8')
```

```python
## print the column names for each dataset
print("Brightest Stars Column Names:", brightest_stars.dtype.names)
print("Nearest Stars Column Names:", nearest_stars.dtype.names)
print("NGC 2477 Column Names:", ngc2477.dtype.names)
print("NGC 2682 Column Names:", ngc2682.dtype.names)
```

```python
## plot the data on an HR diagram
## to make the HR diagram, we need to plot the apparent magnitude against the colour index (bp-rp)

plt.figure(figsize=(10, 6))
plt.scatter(brightest_stars['GAIA_colour_bp__rp'], brightest_stars['GAIA_g_apparent_mag'], label='Brightest Stars')
plt.xlabel('GAIA colour')
plt.gca().invert_yaxis() # invert y-axis to have brighter stars at the top
plt.ylabel('GAIA g apparentmag')
plt.title('HR Diagram - Brightest Stars')
plt.legend()
plt.show()
```
